```python
from pathlib import Path
import numpy as np, pandas as pd, json
import matplotlib.pyplot as plt
from IPython.display import display

OUTPUT=Path('/kaggle/working/nids_outputs/05_V10_Checkpoint_Audit')
OUTPUT.mkdir(parents=True,exist_ok=True)
V10_INPUT_DIR=None  # Ví dụ: '/kaggle/input/your-v10-output/nids_outputs/05_V10_SMOTE_Stability'
required=['validation_bias_grid.csv','validation_best_per_seed.csv','method_seed_stability.csv','all_validation_trials.csv']
if V10_INPUT_DIR:
    folder=Path(V10_INPUT_DIR)
    assert folder.is_dir(),folder
else:
    roots=[Path('/kaggle/input'), Path('/kaggle/working/nids_outputs'),Path('/mnt/data')]
    matches=[]
    for root in roots:
        if root.exists():
            matches.extend(p.parent for p in root.rglob('validation_bias_grid.csv') if '05_V10_SMOTE_Stability' in str(p.parent))
    matches=sorted(set(matches),key=str)
    if not matches:
        raise FileNotFoundError('Không tìm thấy output V10. Hãy Add Input output notebook V10 rồi đặt V10_INPUT_DIR nếu cần.')
    if len(matches)>1:
        print('Có nhiều bản V10:',*[str(p) for p in matches],sep='\n - ')
        raise RuntimeError('Đặt V10_INPUT_DIR để chọn một output V10 cụ thể; tránh trộn lần chạy.')
    folder=matches[0]
assert all((folder/x).exists() for x in required),f'Thiếu file V10 trong {folder}'
grid=pd.read_csv(folder/required[0]); per_seed=pd.read_csv(folder/required[1]); stab=pd.read_csv(folder/required[2]); trials=pd.read_csv(folder/required[3]);
for col in ['strategy','seed','bias','macro_f1','accuracy','xss_recall','xss_f1','fpr']:
    assert col in grid.columns,f'Missing column {col}'
assert len(grid)==60 and grid.groupby(['strategy','seed']).size().eq(5).all(),f'Số hàng không như dự kiến: {len(grid)}'
print('V10:',folder,' | rows:',len(grid),' | methods:',grid.strategy.unique().tolist())
display(grid.head())
```

## 1. Kiểm tra lại thống kê qua 3 seeds
**Phân biệt:** `best per seed` dùng *bias tối ưu Macro-F1* cho từng seed; không phải checkpoint có XSS cao nhất. Recall cao đơn lẻ không bảo đảm F1 cao.

```python
display(stab[['strategy','n_seeds','macro_f1_mean','macro_f1_std','accuracy_mean','xss_recall_mean','xss_f1_mean','fpr_mean']].sort_values('macro_f1_mean',ascending=False).round(4))
print('XSS metric tại checkpoint+threshold có Macro-F1 cao nhất mỗi seed:')
display(per_seed[['strategy','seed','bias','macro_f1','accuracy','xss_recall','xss_f1','fpr']].sort_values(['strategy','seed']).round(4))
```

## 2. Pareto frontier trên toàn bộ validation grid

Một cấu hình nằm trên Pareto frontier nếu không có cấu hình khác **không kém hơn** ở cả ba mục tiêu: **Macro-F1 ↑, XSS F1 ↑, FPR ↓**, đồng thời tốt hơn ít nhất một mục tiêu. Không có ngưỡng tối ưu duy nhất; chỉ dùng Pareto để trình bày các đánh đổi.

```python
def nondominated(df,cols=('macro_f1','xss_f1','fpr')):
    vals=df[list(cols)].to_numpy(float).copy(); vals[:,2]*=-1  # FPR thấp là tốt
    keep=np.ones(len(vals),dtype=bool)
    for i in range(len(vals)):
        better_or_equal=np.all(vals>=vals[i]-1e-12,axis=1)
        strictly_better=np.any(vals>vals[i]+1e-12,axis=1)
        better_or_equal[i]=False
        if np.any(better_or_equal & strictly_better):keep[i]=False
    return keep
pareto=grid.loc[nondominated(grid)].sort_values(['macro_f1','xss_f1'],ascending=False)
pareto.to_csv(OUTPUT/'validation_pareto_frontier.csv',index=False)
display(pareto[['trial_id','strategy','seed','bias','macro_f1','accuracy','xss_recall','xss_f1','fpr']].round(4))
fig,ax=plt.subplots(figsize=(9,6))
for name,part in grid.groupby('strategy'):
    ax.scatter(part.xss_f1*100,part.macro_f1*100,label=name,s=34,alpha=.7)
ax.scatter(pareto.xss_f1*100,pareto.macro_f1*100,marker='x',s=90,label='Pareto')
ax.set_xlabel('Validation XSS F1 (%)');ax.set_ylabel('Validation Macro-F1 (%)');ax.legend();ax.grid(alpha=.25)
fig.tight_layout();fig.savefig(OUTPUT/'validation_tradeoff.png',dpi=160);plt.show()
```

## 3. Phân tích lựa chọn có ràng buộc XSS F1 và FPR

Đây là **phân tích độ nhạy hậu nghiệm**, không tuyên bố cấu hình này đã được chỉ định trước thực nghiệm. Có thể dùng để lên kế hoạch cho *protocol mới trên validation độc lập*, không được dùng test cũ để tinh chỉnh.

```python
scenarios=[
 ('Macro-F1 tối đa',None,None),
 ('XSS F1 >= 20%, FPR <= 10%',.20,.10),
 ('XSS F1 >= 35%, FPR <= 10%',.35,.10),
 ('XSS F1 >= 45%, FPR <= 10%',.45,.10),
 ('XSS F1 >= 50%, FPR <= 7%',.50,.07),
]
selected=[]
for desc,min_xss,max_fpr in scenarios:
    subset=grid.copy()
    if min_xss is not None:subset=subset[subset.xss_f1>=min_xss]
    if max_fpr is not None:subset=subset[subset.fpr<=max_fpr]
    if subset.empty:
        selected.append({'scenario':desc,'feasible':False});continue
    winner=subset.sort_values(['macro_f1','xss_f1','accuracy'],ascending=False).iloc[0]
    selected.append({'scenario':desc,'feasible':True,**winner.to_dict()})
scenarios_df=pd.DataFrame(selected);scenarios_df.to_csv(OUTPUT/'validation_scenarios.csv',index=False)
cols=['scenario','feasible','trial_id','strategy','seed','bias','macro_f1','accuracy','xss_recall','xss_f1','fpr']
display(scenarios_df.reindex(columns=cols).round(4))
```

```python
decision={
 'source_v10_dir':str(folder),
 'validation_configs':int(len(grid)),
 'pareto_count':int(len(pareto)),
 'chosen_by_original_v10_rule':str(stab.sort_values('macro_f1_mean',ascending=False).iloc[0]['strategy']),
 'best_validation_macro_f1':float(grid.macro_f1.max()),
 'best_validation_xss_f1':float(grid.xss_f1.max()),
 'scenario_results':selected,
 'caveat':'Exploratory post-hoc analysis, no test-based selection or independent evaluation',
}
(OUTPUT/'audit_summary.json').write_text(json.dumps(decision,ensure_ascii=False,indent=2,default=str),encoding='utf8')
print(json.dumps({k:v for k,v in decision.items() if k!='scenario_results'},indent=2,ensure_ascii=False))
print('Saved:',OUTPUT, 'files:',[p.name for p in OUTPUT.iterdir()])
```
